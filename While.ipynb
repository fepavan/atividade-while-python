```python
#Como funciona o WHILE?
#WHILE condição:

#Exemplo 1: Contar até onde o usuário pedir.

limite = int(input("Informe um nímero limite: "))
inicio = 0

while inicio <= limite:
  print(inicio)
  inicio = inicio + 1
```

```python
#Controle com WHILE

senha_correta = "abc123"
tentativas = 0

while tentativas < 3:
  senha = input("Digite sua senha: ")
  if senha_correta == senha:
    print("Bem vindo ao sistema")
    break #quebra
  else:
    print("Senha incorreta")
    tentativas = tentativas + 1
    print(f"Número de tentativas = {tentativas}")
    if tentativas == 3:
      print("Senha bloqueada")
```

```python
#WHILE TRUE

while True:
  comando = input("Para sair, digite SAIR: ")
  if comando.lower() == "sair":
    print("Até mais!")
    break
  else:
    print("Continuando o sistema")
```

```python
#Atividade 1 - Fácil

#Sistema que pergunte ao usuário se quer salvar uma informação em uma lista de carros
#A lista --> carros = ["uno","corolla","fusca"]
#Se usuário quer colocar um novo carro, vai adicionar.
#Se não, digitar sair, finaliza o sistema.

carros = ["uno","corolla","fusca"]

print("\n--- Sistema de Gestão de Carros ---")

while True:
  novo_carro = input("\nInforme o nome do novo carro ou digite SAIR para finalizar. ")
  if novo_carro == "sair":
    print("Sistema finalizado. Até logo!")
    break
  else:
      carros.append(novo_carro)
      for i in carros:
        print(i)
```
